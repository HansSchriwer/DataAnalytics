## 2 Capital Asset Pricing Model <a id="Case"></a>
The Capital Asset Pricing Model (CAPM) describes the relationship between returns on a speculative asset (typically returns on stock) and returns on the whole stock market. The underlying theory describes the risk of owning an asset, where risk refers to variation in returns over time. Part of the risk of owning an asset is associated with overall movements in the market as a whole and so is called market risk. The remaining idiosyncratic risk is unique to the asset. The underlying theory promises that markets will pay investors for taking market risk, but not for assuming idiosyncratic risk, which it treats as a gamble.
<br><br>
According to the CAPM, a simple regression model describes the returns $Rt$ on a stock over some time period as:

$$Rt=\alpha+\beta Mt+\epsilon t$$

where Mt is the return on the
market and εt denotes the effects of other factors on the stock return. The intercept is called the α of the stock, and the slope is called the β of the stock. If β = 0, then movements in the market don’t influence the value of the stock. If β=1, then the stock tends to move up and down with changes in the market. Stocks with β>1, sometimes called growth stocks, amplify movements in the market whereas those with β<1 attenuate swings in the market. The claim of CAPM is that investors are not compensated for taking.

### 4.1.2 **CAPM Model**

<b>The precise returns used in the CAPM are so- called excess returns, formed by subtracting the return $Rf$ on a risk-free asset (usually Treasury Bills) from the return on a risky asset.</b>


From difference of the excess returns on Berkshire Hathaway and the stock market, fit the regression 

$$(Rt−Rf)=\alpha′+\beta′ (Mt− Rf)+\epsilon t′$$

Here is the model specification (model2) and the resulting summary statistics for the model.

```python
## Summary Statistics
(returnData["Excess Berkshire Return"] - returnData["Excess Market Return"]).describe()
```

Here is the residual plot for model2.

```python
# RESIDUALS
x_axis = returnData["Excess Market Return"].values
y_axis = returnData["Excess Berkshire Return"].values

sns.residplot(x=x_axis, y=y_axis, data=returnData)
plt.title("Residuals vs Excess Market Return (Per Month)")
plt.xlabel("Excess Market Return (Per Month)")
plt.ylabel("Residuals")
```

In later weeks we will discuss these residual plots in more detail. There are a few things that stand out in this plot:
1. We want the red line (the average conditional error) to be flat and at zero. This curve is pretty good. One could argue that the dip on the left could be due to having very few data points there.
<br><br>
2. We want the spread of the errors to be constant across the graph. This is not terrible, but there  seem to be a lower variance on the right and left sides of the graph than in the middle. However, the middle has fewer data points on the right and left, so the previous result could be the result of the number of data points. Constant variance across values of the explanatory variable (Market.Return) is called homoskedasticity, and it is an important feature of a dataset when doing statistical testing on regression parameters.
