Now let’s look at the residuals from the linear regression. Recall that the residuals are the differences in the actual data and out model’s prediction for each value of crime (or 1/crime). In residual plots we can see all the variation in the output that is NOT explained by the model. If the residuals are all zero, that means that our model was perfect, which is the case only for trivial data sets.
<br><br>
In the residual plots, we want to see that as we move across the values of the x-axis, that the residuals at each value are centered around zero, and that the variation of the residuals is constant across values of x-variable. In statistical parlance this is called “zero conditional mean” and “constant conditional variance”. These are two important assumptions for applying the linear model and using it for statistical tests.
<br><br>
To expand on these assumptions, “zero conditional mean” means that conditioned upon the input variable (meaning move across the values of the x-axis and at each value, hold that constant and look at the average of the residuals for that data (directly above or below the x-axis value). “Constant conditional variance” means that the spread of the data in the y-axis does not vary across the x-axis. The statistical term for this is homoskedasticity for the constant-variance case, and heteroskedasticity for the non-constant case.

### 1.7.4 Residual Plot (Excludes Philly Center City) <a id="Residual"></a>
